# Laptop Data Cleaning Notebook

This notebook prepares the raw laptop dataset for downstream analysis and modeling.

## Section Guide
1. Load and inspect the raw dataset.
2. Clean columns and convert data types.
3. Export the cleaned dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1) Load and Inspect Raw Dataset

Run initial checks for shape, schema, duplicates, and missing values before cleaning.

```python
df = pd.read_csv('data/raw/laptop_data.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['Unnamed: 0'],inplace=True)
```

```python
df.head()
```

## 2) Clean Columns and Convert Data Types

Remove unnecessary columns, strip unit strings (GB, kg), and convert fields to numeric types.

```python
df['Ram'] = df['Ram'].str.replace('GB','')
df['Weight'] = df['Weight'].str.replace('kg','')
```

```python
df.head()
```

```python
df['Ram'] = df['Ram'].astype('int32')
df['Weight'] = df['Weight'].astype('float32')
```

```python
df.info()
```

## 3) Export Cleaned Dataset

After validation, save the cleaned file for visualization and regression notebooks.

```python
df.to_csv('data/processed/cleaned_laptop_data.csv', index=False)
print('Saved data/processed/cleaned_laptop_data.csv')
```
